```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
data = {'Name':['Mouni', 'Sujit', 'Navya', 'Phaninder', 'Sushma', 'Pallabi', 'Nitya', 'Shreya', 'Charan', 'Kalyan'],
        'Age': np.random.randint(18, 25, size=10),
        'Gender':['F', 'M', 'F', 'M', 'F', 'F', 'F', 'F', 'M', 'M'],
        'Salary': np.random.randint(45000, 100000, size=10),
        'Department': ['IT', 'HR', 'HR', 'Finance', 'IT', 'IT', 'IT', 'IT', 'Finance', 'HR'],
        'Joining_Date': pd.date_range(start='2023-01-01', periods=10,freq='MS'),
        'Rating': np.random.uniform(1,5,size=10).round(1)
        }

df = pd.DataFrame(data)
print(df)
```

```python
n=df['Salary'].head(10).mean()
print("The mean of Salary:", n)
```

```python
m=df['Age'].head().mean()
print("The mean of first 5 rows:", m)
```

```python
#Calculate Sum of Salaries and count of employees in each Department
df.groupby('Department').agg({'Name':'count', 'Salary':'sum'})
```

```python
import matplotlib.pyplot as plt
import seaborn as sns  
```

```python
# Calculate average salary per department in bar graph
dept_salary = df.groupby('Department')['Salary'].mean().sort_values(ascending=False)

# Plot
sns.barplot(x=dept_salary.index, y=dept_salary.values,hue=dept_salary.index,palette=['#ff9999','#66b3ff','#ffff00'],legend=False)
plt.title('Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.show() 
```

```python
#Pie Chart: Gender Distribution
# Count gender
gender_count = df['Gender'].value_counts()

# Plot Pie
plt.pie(gender_count, labels=gender_count.index, colors=['#ff9999','#66b3ff'])
plt.title('Gender Distribution')
plt.show()
```

```python
#Box Plot: Salary Distribution by Gender
sns.boxplot(x='Gender', y='Salary', data=df, hue='Gender',palette='Set2',legend=False)
plt.title('Salary Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Salary')
plt.show()
```

```python
#Histogram: Age Distribution
sns.histplot(df['Age'], bins=5, kde=True,color='green')
plt.title('Age Distribution of Employees')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
#Scatterplot: Age v/s Salary
sns.scatterplot(x='Age', y='Salary', data=df, hue='Department', s=100)
plt.title('Age vs. Salary')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.show()
```
